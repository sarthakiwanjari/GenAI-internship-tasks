```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def clean_repeated_chars(word):
    return re.sub(r'(.)\1+', r'\1', word)

def preprocess_text(text):
    if not text or text.strip() == "":
        return [], ""

    # Lowercase
    text = text.lower()

    # Remove URLs & emails
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove emojis & special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    tokens = word_tokenize(text)

    # Handle repeated characters
    tokens = [clean_repeated_chars(word) for word in tokens]

    # Remove short tokens (≤2) except 'no', 'not'
    tokens = [word for word in tokens if len(word) > 2 or word in ['no', 'not']]

    # Remove stopwords (except 'no', 'not')
    tokens = [word for word in tokens if word not in stop_words or word in ['no', 'not']]

    cleaned_sentence = " ".join(tokens)

    return tokens, cleaned_sentence
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

for sentence in test_sentences:
    tokens, clean_sent = preprocess_text(sentence)
    
    print("\nOriginal:", sentence)
    print("Tokens:", tokens)
    print("Cleaned Sentence:", clean_sent)
```

# Task 4: Token Analytics


```python
def analyze_tokens(tokens):
    if len(tokens) == 0:
        return 0, 0, 0
    
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(word) for word in tokens) / total
    
    return total, unique, round(avg_len, 2)

for sentence in test_sentences:
    tokens, _ = preprocess_text(sentence)
    total, unique, avg_len = analyze_tokens(tokens)

    print("\nSentence:", sentence)
    print("Total Tokens:", total)
    print("Unique Tokens:", unique)
    print("Avg Token Length:", avg_len)
```

# Task 5: Frequency Analysis


```python
from collections import Counter

all_tokens = []

for sentence in test_sentences:
    tokens, _ = preprocess_text(sentence)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("Top 10 Frequent Words:", counter.most_common(10))
print("Top 5 Least Frequent Words:", counter.most_common()[:-6:-1])
```

# Task 6: Build Full Pipeline


```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean_sent = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean_sent)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

# Example
result = full_pipeline(test_sentences)
print(result)
```
